```python
from pathlib import Path
import sqlite3
```

```python
current_dir = Path(__file__).parent if "__file__" in locals() else Path.cwd()
db_path = current_dir / "Database" / "carMakeModels.db"

connection = sqlite3.connect(db_path)
c = connection.cursor()
```

```python
print(c.execute("SELECT Fuel_Types FROM cars WHERE Cars_Names = ?", ("SF90 STRADALE",)).fetchone())
```

```python
c.execute("UPDATE cars SET Fuel_Types = ? WHERE Cars_Names = ?",("Petrol", "SF90 STRADALE"))
connection.commit()
print(c.execute("SELECT Fuel_Types FROM cars WHERE Cars_Names = ?", ("SF90 STRADALE",)).fetchone())
```

```python
c.execute("UPDATE cars SET Company_Names = LOWER(Company_Names)")
print(c.execute("SELECT Company_Names from cars").fetchall())
connection.commit()
```
